<b>1. A simple Line

```python
#define line function
def line_func(x):
    return 2*x+1
```

```python
x=np.linspace(-3,3,100)
y=line_func(x)
```

```python
print(x[:5])
print(y[:5])
```

```python
x1=-2
x2=1
y1,y2=line_func(x1),line_func(x2)
print(y1,y2)
```

```python
#choose two points in graph for finding slope
rise=y2-y1
run=x2-x1
slope=rise/run


print("Line example:")
print("point 1:",f"{x1,y1}")
print("point 2:",f"{x2,y2}")
print("Rise:",rise)
print("Runn",run)
print("slope of grpah:",slope)
 
```

```python
#given x and y values find the slopes
plt.figure(figsize=(6,4))
plt.plot(x,y)
plt.title("Linear plot")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(alpha=0.5)
plt.show()
```

Making the curve
y=x^2

```python
def curve(x):
    return x**2

x_vals=np.linspace(-3,3,100)
y_vals=curve(x_vals)

plt.figure(figsize=(6,4))
plt.plot(x_vals,y_vals)
plt.title("quadratic plot plot")
plt.xlabel("X_vals")
plt.ylabel("y_vals")
plt.grid(alpha=0.5)
plt.show()
```

```python
def slope_estimates(func,x1,h=0.00001):
    x2=x1+h
    y1=curve(x1)
    y2=func(x2)
    dy=y2-y1
    dx=h
    slope=dy/dx
    return slope
    
    
```

```python
points=[-2,0,2,4,6,5]
print("Slope estimates for y=x^2 on small change in x")
for p in points:
    s=slope_estimates(curve,p)
    print(f"x={p}-> estimated slope={s:.2f}")
```

<b>Calculus part 2

scenioro 1: when (dl/dw) is positivie
w=w-lr* +ve_number
so w=w-some number

scenario 2: when (dl/dw ) is negative
w=w-lr* -ve_number
w=w+number
  

```python
x=np.array([1,2,3])
y_true=np.array([2,4,6])


#f(x)=w*x
#y=w*x
w=1.0

#learning rate 
lr=0.01

def loss(w):
    y_pred=w*x
    return np.mean((y_pred - y_true)**2)

def dloss(w):
    y_pred=w*x
    return np.mean(2*x*(y_pred - y_true))

print("start w:",w, "loss:",loss(w))
for step in range(30):
    slope=dloss(w)
    w=w-lr*slope
    print("step:",step+1,"w:",round(w,3),"loss:",round(loss(w),3))
```
