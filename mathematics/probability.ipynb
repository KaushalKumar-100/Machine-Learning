<b>1. Coin Toss Simulation

```python
n=500

outcomes=np.random.choice(("H","T"),size=n)


heads=np.sum(outcomes=="H")
tails=np.sum(outcomes=="T")

print("Heads: ",heads)
print("tails: ",tails)
print("Estimated p(heads): ",heads/n)
print("Estimated p(tails): ",tails/n)


 
```

<b>2. Dice Roll Simulation

```python
n=1000
dice=np.random.randint(1,7,size=n)

p_four=np.mean(dice==4)
print("Estimated p(4): ",p_four)

vals,counts=np.unique(dice , return_counts=True)
print(vals)
print(counts)

plt.bar(vals,counts)
plt.title("Dice rolls Outcomes")
plt.xlabel("Number")
plt.ylabel("counts")
plt.show()
```

<b>3. Conditional Probability

We create a tiny datasets of 100 students .Each student may like or not ML

```python
data=pd.DataFrame({
    "likes_Ml": np.random.choice([1,0],100,p=[0.6,0.4])
})

#Dl depends on whether the student like ML
data["likes_dl"]=[np.random.choice([1,0],p=[0.7,0.3]) if ml
                  else
                  np.random.choice([1,0],p=[0.2,0.8])
                  for ml in data["likes_Ml"]
                  ]

p_ml=data["likes_Ml"].mean()
p_dl=data["likes_dl"].mean()
p_dl_given_ml=data[data["likes_Ml"]==1]["likes_dl"].mean()

print("p(ml):",round(p_ml,2))
print("p(dl):",round(p_dl,2))
print("p(p_dl_given_ml):",round(p_dl_given_ml,2))


```

<b>5.Distribution of <u>data<u/>

```python
#visualising the  uniform data distributions
n=5000
plt.figure(figsize=(5,4))
plt.hist(np.random.uniform(0,1,n),bins=20,alpha=0.9,edgecolor="k")
plt.title("Uniform Distribution")
plt.show()

plt.figure(figsize=(5,4))
plt.hist(np.random.normal(0,1,n),bins=20,alpha=0.9,edgecolor="k")
plt.title("Normal Distribution")
plt.show()

plt.figure(figsize=(5,4))
vals,counts=np.unique(np.random.binomial(1,0.3,n),return_counts=True)
plt.bar(vals,counts)
plt.title("Bernouli Distribution")
plt.show()


plt.figure(figsize=(5,4))
vals,counts=np.unique(np.random.binomial(10,0.3,n),return_counts=True)
plt.bar(vals,counts)
plt.title("Binomial Distribution")
plt.show()


```
