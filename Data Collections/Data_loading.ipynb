**5. Compare the file size**


```python
def file_size(path):
    size_bytes=os.path.getsize(path)
    size_kb=round(size_bytes/1024,2)
    size_mb=round(size_kb/1024,2)
    return size_kb,size_mb

for f in ["titanic.csv","titanic.xlsx","titanic.parquet","titanic-tab.csv"]:
    kb,mb=file_size(f)
    print(f"size in kb: {kb} and size in mb: {mb}")
```
