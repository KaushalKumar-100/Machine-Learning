```python
x=df.drop(columns=["Outcome"])
y=df["Outcome"]
```

5 . K fold setup

```python
kf=KFold(n_splits=5,shuffle=True,random_state=42)
```

6 Loop through the folds

```python
for fold,(train_idx,test_idx) in enumerate(kf.split(x),start=1):
    print(fold)
    print(len(train_idx))
    print((train_idx[:5]))
    
    print(len(test_idx))
    print((test_idx[:5]))
    
    print ("-"*50)
    
```

```python
accuracy_list=[]
precision_list=[]
recall_list=[]
f1_list=[]
roc_auc_list=[]
```

```python
for fold, (train_idx,test_idx) in enumerate(kf.split(x),start=1):
    #split data
    
    X_train,X_test=x.iloc[train_idx],x.iloc[test_idx]
    y_train,y_test=y.iloc[train_idx],y.iloc[test_idx]
    
    scaler=StandardScaler()
    X_train_scaled=scaler.fit_transform(X_train)
    X_test_scaled=scaler.transform(X_test)
    
    #training the modell
    model=LogisticRegression()
    
    model.fit(X_train_scaled,y_train)
    
    #getting the inference
    
    y_pred=model.predict(X_test_scaled) 
    y_prob=model.predict_proba(X_test_scaled)
    
    #evalutation metrics
    acc=accuracy_score(y_test,y_pred)
    prec=precision_score(y_test,y_pred)
    rec=recall_score(y_test,y_pred)
    f1=f1_score(y_test,y_pred)
    auc=roc_auc_score(y_test,y_pred)    
    
    #store matrix
    accuracy_list.append(acc)
    precision_list.append(prec)
    recall_list.append(rec)
    f1_list.append(f1)
    roc_auc_list.append(auc)
    
    
    #print fold result
    
    print(f" fold : {fold}")
    print(f"  Accuracy : {acc:.3f}")
    print(f"  precision : {prec:.3f}")
    print(f"  Recall : {rec:.3f}")
    print(f"  f1_score : {f1:.3f}")
    
    print(f"  rec_auc : {auc:.3f}")
    print("-"*50)

    
    
    
    
    
```

```python
print(precision_list)
print(accuracy_list)
```

```python
np.mean(accuracy_list)
```

```python
mean_accuracy=round(np.mean(accuracy_list)*100,3)
mean_precision=round(np.mean(precision_list)*100,3)
mean_recall=round(np.mean(recall_list)*100,3)
mean_f1=round(np.mean(f1_list)*100,3)
mean_roc_auc=round(np.mean(roc_auc_list)*100,3)

```

```python
print("Kfold cross validation:")
print(f"Mean Accuracy: {mean_accuracy}")
print(f"Mean precision: {mean_precision}")
print(f"Mean recall: {mean_recall}")
print(f"Mean f1: {mean_f1}")
print(f"Mean roc_auc: {mean_roc_auc}")
```
