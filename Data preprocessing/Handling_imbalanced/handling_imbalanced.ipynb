```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split


from imblearn.under_sampling import RandomUnderSampler
from imblearn.over_sampling import RandomOverSampler,SMOTE

pd.set_option("display.max_columns",None)
```

```python
df=pd.read_csv("diabetes.csv")
print(df.shape)
```

```python
df.isnull().sum()
```

**Target distributions**

```python
print(df["Outcome"].value_counts())
print(df["Outcome"].value_counts(normalize=True).round(3))
```

Train test split


```python
x=df.drop(columns=["Outcome"])
y=df["Outcome"]
```

```python
X_train, X_test, y_train, y_test=train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
print(X_train.shape)
print(X_test.shape)

```

```python
y_train.value_counts(normalize=True).round(3)
```

Random under Sampler


```python
rus=RandomUnderSampler(random_state=42)
```

```python
X_train_rus,y_train_rus=rus.fit_resample(X_train,y_train)
 
```

```python
print(y_train_rus.value_counts())
```

```python
print(X_train_rus.count())
```

Random over sampler

```python
ros=RandomOverSampler(random_state=42)
X_train_ros, y_train_ros=ros.fit_resample(X_train,y_train)
```

```python
y_train_ros.value_counts()
```

```python
smote=SMOTE(random_state=42)
```

```python
X_train_smote,y_train_smote=smote.fit_resample(X_train,y_train)
```

```python
print(y_train_smote.value_counts())
```

```python
target="Outcome"
```

```python
train_smote=X_train_smote.copy()
train_smote[target]=y_train_smote.values

test_original=X_test.copy()
test_original[target]=y_test.values


train_smote.to_csv("diabetes_train_imbalanced.csv")
test_original.to_csv("diabetes_test_imbalanced.csv")
```
