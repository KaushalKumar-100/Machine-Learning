```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
df=pd.read_csv("diabetes.csv")
```

```python
x=df.drop(columns=["Outcome"])
y=df["Outcome"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(
    x,
    y,
    random_state=42,
    test_size=0.2,
    stratify=y
)
```

```python
nums_cols=X_train.select_dtypes(include=[np.number]).columns.tolist()
cat_cols=[col for col in X_train.columns if col not in nums_cols]
```

```python
nums_cols
```

```python
scaler=StandardScaler()
scaler.fit(X_train[nums_cols])
```

```python
X_train_std=X_train.copy()
X_test_std=X_test.copy()
```

```python
X_train_std[nums_cols]=scaler.transform(X_train[nums_cols])
X_test_std[nums_cols]=scaler.transform(X_test[nums_cols])
```

```python
print(X_train.describe().T)
```

```python
print(X_train_std.describe().T)
```

5. Normalization -MinmaxScaler

```python
mm=MinMaxScaler()
```

```python
mm.fit(X_train[nums_cols])
```

```python
X_train_mm=X_train.copy()
X_test_mm=X_test.copy()

X_train_mm[nums_cols]=mm.transform(X_train[nums_cols])
X_test_mm[nums_cols]=mm.transform(X_test[nums_cols])

```
