```python
import pandas as pd
from sklearn.model_selection import train_test_split
import seaborn as sns
pd.set_option("display.max_columns",None)
```

# load dataset

```python
df=sns.load_dataset("titanic")
```

```python
df.head()
```

```python
df.columns.tolist()
```

```python
# dropping alive column as it is redunant due to presense of survived columns
df=df.drop(columns=["alive"])
```

```python
df.shape
```

 3. Drop row with missing target( safe before train test split)

```python
df.isnull().sum()
```

```python
target="survived"

missing_target_count=df[target].isna().sum()
missing_target_count

#drop the row that has missing target value
df=df.dropna(subset=[target])
df.shape
```

4. Separate x(featues) and y(target)  

```python
x=df.drop(columns=[target]) 
y=df[target]
```

```python
print(x.shape) 
print(y.shape)
```

```python
x[:5]
```

```python
y[:5]
```

```python
print(y.value_counts())
print(y.value_counts(normalize=True).round(3))
```

**Train test and split**

```python
X_train, X_test, y_train, y_test=train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
print("shape:")
print("X_train",X_train.shape)
print("X_test",X_test.shape)
print("y_train",y_train.shape)
print("y_test",y_test.shape)
```

```python
X_train.head()
```

```python
y_train.head()
```

X_train, y_train --> Ml model learn the pattern

X_test, y_test --> Ml model Evalution

```python
print("Target distribution of full data: ")
print(y.value_counts())

print("-"*50)
print("Target distributions on training data")
print(y_train.value_counts())

print("-"*50)
print("target distributions in percentage")
print(y_train.value_counts(normalize=True).round(3))


print("-"*50)
print("Target distributions on test data")
print(y_test.value_counts())

print("-"*50)
print("target distributions in percentage")
print(y_test.value_counts(normalize=True).round(3))
print("-"*50)

```

6.Save Splits

```python
train_df=X_train.copy()
train_df[target]=y_train.values
```

```python
train_df.sample(6)
```

```python
test_df=X_test.copy()
test_df[target]=y_test.values
```
