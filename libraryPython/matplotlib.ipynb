```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x=np.linspace(0,10,50)
print(len(x))
print(x)
```

```python
y=np.sin(x)
plt.plot(x,y)
plt.title("Simple line plot")
plt.xlabel("x-axis")
plt.ylabel("Y=sin(x)")
plt.show()
```

<h4>Customized graph<h4/>

```python
plt.figure(figsize=(7,4))
plt.plot(x,y,color="red",linestyle="--",label="sin(x)")
plt.title("customise line graph")
plt.xlabel("x-axis")
plt.ylabel("y=sin(x)")
plt.legend()
plt.grid(True)
plt.show()
```

<h3>Bar graph<h3/>

```python
students=["Alice","bob","charlie","dice"]
scores=[45,34,23,66]
plt.figure(figsize=(7,5))

plt.bar(students,scores,color="orange")
plt.title("student scores")
# plt.label("scores")
plt.xlabel("student")
plt.ylabel("scores")
```

```python
students=["Alice","bob","charlie","dice"]
scores=[45,34,23,66]
plt.figure(figsize=(7,5))

plt.barh(students,scores,color="orange")
plt.title("Horizontal bar chart student scores")
# plt.label("scores")
plt.xlabel("student")
plt.ylabel("scores")
```

```python
np.random.seed(10)
x=np.random.rand(50)
y=np.random.rand(50)

print(x[:5])
print(y[:5])
```

```python
plt.scatter(x,y,color="purple",alpha=1)
plt.title("scatter plot")
plt.xlabel("Feature1")
plt.ylabel("Featute2")
plt.show()
```

<h3>Histogram<h3/>

```python
data=np.random.randn(1000)
plt.hist(data,bins=30,color="blue",edgecolor="black")
plt.title("histogram show")
plt.xlabel("feature1")
# plt.ylable("feature2")
plt.legend()
plt.show()
```

***Introduction to sub-Plot ***


```python
x=np.linspace(0,10,100)

plt.figure(figsize=(10,4))

plt.subplot(2,1,1)
plt.plot(x,np.sin(x),"r--")
plt.title("sine")

plt.subplot(2,1,2)
plt.plot(x,np.cos(x),"b--")
plt.title("Cosine")

plt.suptitle("sine and cosine curve")
plt.show()
```
