```python
#set some global variables
sns.set_style("darkgrid")
sns.set_palette("viridis")
```

```python
#histogram -Distribution of total bill
plt.figure(figsize=(6,4))

sns.histplot(tips["total_bill"],kde=True,color="green")
plt.title("total billls")
plt.xlabel("Total Bills")
plt.ylabel("count")
plt.show()
```

Boxplot

```python
plt.figure(figsize=(6,4))

sns.boxplot(x="day",y="total_bill",data=tips,legend=False)
plt.title("total_bills")
plt.show()
```

<h2>count-Plot

```python
plt.figure(figsize=(6,4))

sns.countplot(x="day",data=tips,hue="sex")
plt.title("total_bills day")
plt.show()
```

```python
sns.pairplot(tips,hue="sex",diag_kind="kde") 
plt.suptitle("Pair -wise relationship")
plt.show()
```

heatmap

```python
correlation_matrix=tips.corr(numeric_only=True)
plt.figure(figsize=(6,4))
sns.heatmap(correlation_matrix,annot=True,cmap="coolwarm",linewidth=0.5)
plt.title("Feature Correlation matrix")
plt.show()
```
