4.4 Data type Validation

```python
expected_types={
    "survived":"int64",
    "pclass":"int64",
    "sex":"category",
    "age":"float",
    "fare":"float64"
}
# print(expected_types)
for col,expected in expected_types.items():
    if col  in df.columns:
        actual=df[col].dtypes
        print(f"{col}: actual={actual},expected:{expected}")
```

4.5 Constant and quasi constant columns

```python
df.nunique()
```

```python
n_rows=len(df)
nunique=df.nunique()

constant_cols=nunique[nunique==1].index.to_list()
# print(constant_cols) 

#quasi constant column
#95 % percent of value is common
quasi_constant_cols=[]
for col in df.columns:
    top_freq=df[col].value_counts(normalize=True, dropna=False).values[0]
    if top_freq>0.75 and col not in constant_cols:
        quasi_constant_cols.append(col)
        
        
        
print("Quasi constant columns (top value more than 90 % ):\n ",quasi_constant_cols)

```

```python
df["pclass"].value_counts().values
```

Id like columns

columns wherer number of unique value is near to number of column

```python
n_rows=len(df)

id_like_cols=[]

for col in df.columns:
    if df[col].nunique(dropna=False)==n_rows:
        id_like_cols.append(col)
        
        
print("Id like colums: ",id_like_cols)
```

4.9 High zero columns (numeric)

```python
numeric_cols=df.select_dtypes(include=[np.number]).columns.to_list() 
zero_share={}
for col in numeric_cols:
    zero_share[col]=(df[col]==0).mean()
    
zero_share_series=pd.Series(zero_share).sort_values(ascending=False)

print(zero_share_series )
print("--"*30)

high_zero_threshold=0.7
high_zero_cols=zero_share_series[zero_share_series >=high_zero_threshold]
print("Numeric columns with many zeros:",high_zero_cols)


 
```
